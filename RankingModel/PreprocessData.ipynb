```python
import torch 
label_cols = ['verified_purchase','clicked','add_to_cart','favorite']
label_counts = df[label_cols].sum()
total_counts = len(df)
neg_counts = total_counts - label_counts

print(f"label_counts: {label_counts}, total_counts: {total_counts}, neg_counts: {neg_counts}")

# Compute pos_weight = neg / pos
pos_weight = (neg_counts / label_counts).values
print("/n Pos weights:", pos_weight)  # Optional debug
pos_weight_tensor = torch.tensor(pos_weight, dtype=torch.float32)

print(f"/n pos_weight_tensor: {pos_weight_tensor}")
```
